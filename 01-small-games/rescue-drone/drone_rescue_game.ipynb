```python
import random


# ------------------------------------------------
# -            LOCATION DATA                     -
# ------------------------------------------------

parking_lot = {"name" : "Parking Lot", "distance" : 6, "risk" : "low", "people" : 3}
campus_center = {"name" : "Campus Center", "distance" : 10, "risk" : "medium", "people" : 5}
unity_hall = {"name" : "Unity Hall", "distance" : 15, "risk" : "high", "people" : 7}
labs = {"name" : "Labs", "distance" : 20, "risk" : "super_high", "people" : 10}
base = {"name" : "Base", "distance" : 0, "risk" : "none", "people" : 0}

locations = [parking_lot, campus_center, unity_hall, labs, base]

# ------------------------------------------------
# -            STARTING VARIABLES                -
# ------------------------------------------------
game_over = False
max_battery = 100
battery = 100
distance = 0
current_location = "Base"
people_onboard = 0
people_saved = 0
battery_per_km = 5
rescue_cost = 5
rolls_available = 0
rescue_stopped = False
multi_roll_unlocked = False

# ------------------------------------------------
# -                FUNCTIONS                     -
# ------------------------------------------------
def instructions():
  instructions_title = " INSTRUCTIONS "
  print(f"\n{instructions_title.center(40, '*')}")
  print("\n 1. IF YOU RUN OUT OF BATTERY TO RETURN TO BASE YOU LOSE")

def show_menu():
  available_zones_title = " AVAILABLE ZONES "
  print(f"\n{available_zones_title.center(40, '*')}")
  print("1. Parking Lot")
  print("2. Campus Center")
  print("3. Unity Hall")
  print("4. Labs")
  print("5. Return to base")


def show_details(destination):
  location_details_title = "LOCATION DETAILS "
  print(f"\n{location_details_title.center(40, "*")}")
  print(f"Location: {destination['name']}")
  print(f"Distance: {destination['distance']} km")
  print(f"Risk: {destination['risk'].replace("_", " ").upper()}")
  print(f"People detected: {destination['people']}")
  print(f"Each km will consume {battery_per_km}% of your battery")


def battery_cost(destination):
  distance = destination['distance']
  return distance * battery_per_km


def rescue_attempt(destination):
  risk = destination['risk']
  if risk == "low":
    success_rate = 80
  if risk == "medium":
    success_rate = 65
  if risk == "high":
    success_rate = 40
  if risk == "super_high":
    success_rate = 20
  return success_rate

def ROLL():
  global max_battery
  global rescue_cost
  global battery_per_km

  upgrade_roll = random.randint(1, 100)
  print(f"DEBUG upgrade_roll = {upgrade_roll}")
  if upgrade_roll <= 80:
    ammount_roll = random.randint(1,6)
    if ammount_roll in range(1,4):
      max_battery += 5
      print(f"+ 5% max battery!")
      print(f"Battery increased to {max_battery}%")
      return
    elif ammount_roll in range(4,6):
      max_battery += 10
      print(f"+ 10% max battery!")
      print(f"Battery increased to {max_battery}%")
      return
    elif ammount_roll == 6:
      max_battery += 15
      print(f"+ 15% max battery!")
      print(f"Battery increased to {max_battery}%")
      return

  elif 80 < upgrade_roll <= 90:
    ammount_roll = random.randint(1,2)
    if ammount_roll == 1:
      battery_per_km -= 1
      print(f"- 1 battery cost per km!")
      print(f"Now it costs {battery_per_km}%")
      return
    elif ammount_roll == 2:
      rescue_cost -= 1
      print(f"- 1 rescue cost!")
      print(f"Now it costs {rescue_cost}%")
      return
  elif upgrade_roll > 90:
    max_battery +=30
    print(f"+ 30% max battery!")
    print(f"Battery increased to {max_battery}%")
    return

  return


def return_to_base():
  global people_saved
  global people_onboard
  global battery
  global rolls_available
  global rescue_cost
  global current_location
  global multi_roll_unlocked

  current_location = base["name"]
  rescue_cost = 5
  BASE = " BASE "
  print(f"\n{BASE.center(40, '*')}")
  battery = max_battery
  print(f"BATTERY CHARGED TO {battery} %")
  people_saved += people_onboard
  rolls_available += people_onboard
  rolls_reference_num = rolls_available
  people_onboard = 0
  print(f"You have saved {people_saved} people")
  if people_saved == 8 and multi_roll_unlocked == False:
    print("\n*** UPGRADE UNLOCKED ***")
    print("  RESCUE MULTI-ROLL ")
    print("\n Now you can choose how many rescue attempts you ran at once!")
    multi_roll_unlocked = True

  print(f"and obtained {rolls_available} ROLLS")
  MENU = " MENU "
  print(f"\n{MENU.center(40, '*')}")
  print("1. Available zones")
  print(f"2. ROLL ALL (you have {rolls_available} rolls)")
  print("3. Stats")
  choice_input = input("\nChoose your option: ")
  if choice_input.isdigit():
    choice = int(choice_input)
    if choice == 1:
      show_menu()
    elif choice == 2:
      roll_count = 1
      while rolls_available > 0:

        print (f"\nROLL {roll_count} / {rolls_reference_num}")
        print(".........................................")
        ROLL()
        print(".........................................")
        battery = max_battery
        rolls_available -= 1
        roll_count += 1

        if rolls_available == 0:
          break

        roll_again = input("Press enter to roll again ")
        if roll_again!= "":
          break


    elif choice == 3:
  # will be finished later
      print("STATS")
    else:
      print("Invalid option.")
  else:
    print("Invalid input. Please enter a number.")


# ------------------------------------------------
# -               MAIN PROGRAM                   -
# ------------------------------------------------
title = " RESCUE DRONE SYSTEM "
print(title.center(40, "*"))

while game_over == False:
  show_menu()
  choice_destination_input = input("\nChoose destination: ")

  if choice_destination_input.isdigit():
    choice_destination = int(choice_destination_input)

    # choice destination
    if choice_destination >= 1 and choice_destination <= 5:
      if choice_destination == 5:
        return_to_base()
        continue
      destination = locations[choice_destination - 1]
      location = locations[choice_destination - 1]["name"]

      if location == current_location:
        print(f"\nYOU ARE ALREADY AT {location.upper()}")



      # battery consuption and confirmation
      if location != current_location:
        show_details(destination)
        cost = battery_cost(destination)
        print(f"\nBattery aviable: {battery}%")
        print(f"Battery requiered: {cost}%")
        confirm_destination = input("Type enter to confirm, type anything to return to the menu ")

        if confirm_destination != "":
          continue


        if battery >= cost:
          print("\nTravel authorized")
          battery = battery - cost
          print(f"Battery remaining: {battery}%")
          current_location = location
          print(f"You are now at: {current_location}")

        else:
          print("ERROR : not enough battery to travel to destination.")
          continue  # Go back to menu if not enough battery

      else:
        print(f"\nYOU ARE ALREADY AT {location.upper()}")



        # rescue srarts here
      MISSION_STARTED = (" MISSION STARTED ")
      print(f"\n{MISSION_STARTED.center(40, '*')}")

      if destination["people"] == 0:
        print("\nAll survivors at this location have already been rescued.")
        continue # Go back to main menu if no people to rescue

      success_rate = rescue_attempt(destination)


      print("\n*** Note: each survivor increases")
      print("battery consumption per rescue attempt by a point")
      print("and battery consumption per km by 2 point")


      # 1. calculate rescue_cost
      survivors_available = destination['people']
      reference_survivors_num = survivors_available
      rescued_here = 0

      while  survivors_available > 0 and game_over == False:

        total_rescue_cost = rescue_cost + people_onboard
        return_base_cost = destination['distance'] * battery_per_km + 2 * people_onboard

        print("\n--------------------------------------")
        print(f"Sucess rate = {success_rate}%")
        print(f"Rescue cost = {total_rescue_cost}%")
        print(f"Battery remaining = {battery}%")
        print(f"Battery needed to return to base = {return_base_cost}% ")
        print("--------------------------------------\n")
        print(f"SURVIVOR {rescued_here + 1}")


        # --------------------------------
        # CHOOSE NUMBER OF ATTEMPTS
        # --------------------------------

        if multi_roll_unlocked == False:
          attempts = 1
          print("Type anything to stop")

          command = input("Enter to attempt ")

          # 3. if player stops -> break
          if command != "":
            print("\n RESCUE STOPPED")
            rescue_stopped = True
            break


        else:
          attempts_input = input("How many rescue attempts? ")
          if attempts_input.isdigit(): #
            attempts = int(attempts_input)
          else:
            print("Invalid input. Please enter a number.")
            continue


          if attempts == 0:
            print("\n RESCUE STOPPED")
            rescue_stopped = True
            break


        # --------------------------------
        # PERFORM ATTEMPTS
        # --------------------------------

        print('.' * 40)

        for attempt_number in range(attempts):


          if survivors_available == 0:
            break

          total_rescue_cost = rescue_cost + people_onboard
          return_base_cost = destination['distance'] * battery_per_km + 2 * people_onboard


          if (battery - total_rescue_cost ) >= return_base_cost:
            roll = random.randint(1, 100)
            battery -= total_rescue_cost


            if roll <= success_rate:
              people_onboard += 1
              rescued_here += 1
              destination['people'] -= 1
              survivors_available -= 1

              SAVED =f" SAVED {rescued_here} / {reference_survivors_num} "
              print(f"\n{SAVED.center(40,'.')} \n")


            else:
              FAILED = (f" FAILED {rescued_here} / {reference_survivors_num} ")
              print(f"\n{FAILED.center(40,'.')} \n ")

            print('.' * 40)

          else:
            print("\n...........................................................")
            print("!!!!!!!!! NOT ENOUGH BATTERY TO RETURN TO BASE !!!!!!!!!")
            print("!!!!!!!!!             GAME OVER                !!!!!!!!!")
            print("...........................................................")
            game_over = True
            break


        if not game_over and not rescue_stopped and survivors_available == 0:

          if survivors_available == 0:
            print(f"\n ALL SURVIVORS SAVED")

            print(f"\nSURVIVORS ONBOARD = {people_onboard}")
            print(f"Battery remaining = {battery}%")
            print(f"Return to base?")
            print("1. Yes")
            print("2. No")

            return_choice_input = input("\nChoose option: ")
            if return_choice_input.isdigit():
              return_choice = int(return_choice_input)

              if return_choice == 1:
                  return_to_base()

              elif return_choice == 2:
                  continue

              else:
                  print("Invalid option")
                  continue  # If invalid, go back to main menu
          else:
            print("Invalid input. Please enter a number.")
            continue

    else:  # This else belongs to `if choice_destination >= 1 and choice_destination <= 5:`
        print("Invalid option. Please choose a number between 1 and 5.")
        continue

  else:
    print("Invalid input. Please enter a number between 1 and 5.")
    continue
```
